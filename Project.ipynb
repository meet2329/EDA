```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#df=pd.read_csv("")
df = pd.read_csv("stress_data.csv")
df.columns = [
    "Timestamp",
    "Age",
    "Gender",
    "Study_Hours",
    "Class_Attendance",
    "Tuition",
    "Exam_Frequency",
    "Assignment_Load",
    "Sleep_Hours",
    "Physical_Exercise",
    "Social_Media_Use",
    "Screen_Time",
    "Family_Income_Level",
    "Peer_Pressure",
    "Family_Support",
    "Anxiety_Level",
    "University_Type",
    "Stress_Score",
    "Stress_Level"
]
df = df.drop("Timestamp", axis=1)
#df.head()
#df.tail()
'''
We imported the required libraries and loaded the stress dataset.
We renamed the columns for better understanding and removed the unnecessary Timestamp column.
'''
```

```python
#1. View Data
# -------------------------------
df.head()        # First 5 rows
df.tail()        # Last 5 rows
print(df.shape)

# We used head() and tail() to view the first and last rows of the dataset.

# 2. Dataset Information
# -------------------------------
print("Dataset Info:")
df.info()

#We used info() to check the columns, data types, and number of entries.

# 3. Statistical Summary
# -------------------------------
print("\nStatistical Summary:")
print(df.describe())

#We used describe() to understand the basic statistics of numerical data.

# 4. Missing Values Check
# -------------------------------
print("\nMissing Values:")
print(df.isnull().sum())

#We checked for missing values to ensure the dataset is complete.

# 5. Duplicate Check
# -------------------------------
print("\nDuplicate Rows:")
print(df.duplicated().sum())

#We checked for duplicate rows to avoid repeated data affecting our analysis.
```

```python
print(df["Gender"].value_counts())

#We used value_counts() to check the number of students in each gender category.

'''
# Check data types
print(df.dtypes)
'''
# Check unique values
for col in df.columns:
    print("\n", col)
    print(df[col].unique())
    
#We checked unique values in each column to understand the different categories present in the dataset.
    
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numeric Columns:")
print(numeric_columns.tolist())

#We identified all numerical columns in the dataset for further statistical analysis and visualization.
```

```python
categorical_columns = df.select_dtypes(include="object").columns

print("Categorical Columns:")
print(categorical_columns.tolist())

#We identified all categorical columns to analyze different groups and categories in the dataset.
```
